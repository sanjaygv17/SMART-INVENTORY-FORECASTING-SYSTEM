```python
def predict_category(category_name):

    # Filter category data
    category_data = category_weekly[
        category_weekly["Category"]
        == category_name
    ]

    # Get latest row
    latest_row = category_data.iloc[-1]

    # Encode category
    encoded_category = (
        category_encoder.transform(
            [category_name]
        )[0]
    )

    # Create input dataframe
    input_data = pd.DataFrame({
        "Category_Encoded":
        [encoded_category],

        "Week":
        [latest_row["Week"]],

        "Weekly_Revenue":
        [latest_row["Weekly_Revenue"]],

        "Weekly_Margin":
        [latest_row["Weekly_Margin"]],

        "Lag_1":
        [latest_row["Lag_1"]],

        "Lag_2":
        [latest_row["Lag_2"]],

        "Lag_3":
        [latest_row["Lag_3"]],

        "Rolling_4_Week_Avg":
        [latest_row[
            "Rolling_4_Week_Avg"
        ]]
    })

    # Predict
    prediction = (
        category_model.predict(
            input_data
        )[0]
    )

    print(
        f"Predicted next week demand for {category_name}: "
        f"{round(prediction)} units"
    )
```

```python
def predict_product(product_name):

    # Filter product data
    product_data = product_weekly[
        product_weekly["Product_Name"]
        == product_name
    ]

    # Check if product exists
    if len(product_data) == 0:
        print("Product not found!")
        return

    # Get latest row
    latest_row = product_data.iloc[-1]

    # Encode product
    encoded_product = (
        product_encoder.transform(
            [product_name]
        )[0]
    )

    # Encode category
    encoded_category = (
        category_encoder.transform(
            [latest_row["Category"]]
        )[0]
    )

    # Encode brand
    encoded_brand = (
        brand_encoder.transform(
            [latest_row["Brand"]]
        )[0]
    )

    # Create input dataframe
    input_data = pd.DataFrame({
        "Product_Encoded":
        [encoded_product],

        "Category_Encoded":
        [encoded_category],

        "Brand_Encoded":
        [encoded_brand],

        "Week":
        [latest_row["Week"]],

        "Month":
        [latest_row["Month"]],

        "Weekly_Revenue":
        [latest_row["Weekly_Revenue"]],

        "Weekly_Margin":
        [latest_row["Weekly_Margin"]],

        "Average_Stock":
        [latest_row["Average_Stock"]],

        "Lag_1":
        [latest_row["Lag_1"]],

        "Lag_2":
        [latest_row["Lag_2"]],

        "Lag_3":
        [latest_row["Lag_3"]],

        "Lag_4":
        [latest_row["Lag_4"]],

        "Rolling_4_Week_Avg":
        [latest_row[
            "Rolling_4_Week_Avg"
        ]],

        "Rolling_8_Week_Avg":
        [latest_row[
            "Rolling_8_Week_Avg"
        ]]
    })

    # Predict
    prediction = model.predict(
        input_data
    )[0]

    print(
        f"Predicted next week demand for "
        f"{product_name}: "
        f"{round(prediction)} units"
    )
```
